```python
import pandas as pd
import json
```

```python
path = '../data/weather_data.json'
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)
```

```python
with open(path) as f:
    data = json.load(f) #Transformar dict objeto python

df = pd.json_normalize(data)

df.head()
```

```python
timestamp = df["dt"]
offset = df["timezone"]


from datetime import datetime, UTC, timedelta,timezone

local_dt = datetime.fromtimestamp(timestamp, tz=UTC).astimezone(timezone(timedelta(seconds=offset)))
```
